### 1. Scipy's interpolate methods to interpolate HMI:

First, some tutorials on generic interpolation:

```python
from scipy import interpolate as inter
```

```python
n_coarse = 8

x_coarse = np.linspace(1, 100, n_coarse)
y_coarse = np.linspace(1, 100, n_coarse)

X_coarse, Y_coarse = np.meshgrid(x_coarse, y_coarse, indexing='ij')


T = 10
Z = np.cos(X_coarse / T) * np.sin(Y_coarse //T)


plt.figure(figsize = [12, 8])
plt.subplot(1,2,1)
plt.title('Original (coarse) data')
plt.imshow(Z, extent=[x_coarse[0], x_coarse[-1], y_coarse[0], y_coarse[-1]], origin='lower')

scale_factor = 10

x_fine = np.linspace(1, 100, n_coarse * scale_factor)
y_fine = np.linspace(1, 100, n_coarse * scale_factor)

X_fine, Y_fine = np.meshgrid(x_fine, y_fine, indexing='ij')

# This method NEEDS a regular grid to interpolate onto
interpolator_setup = inter.RegularGridInterpolator((x_coarse, y_coarse), Z, method='cubic')
Z_fine = interpolator_setup((X_fine, Y_fine))

plt.subplot(1,2,2)
plt.title(f'Interpolated (upscaled by factor of {scale_factor}) grid')
plt.imshow(Z_fine, extent=[x_coarse[0], x_coarse[-1], y_coarse[0], y_coarse[-1]], origin='lower')
```

```python
#########################################################################
# What if we have an UNEVEN grid of points? Here, logarithmically spaced:
#########################################################################

n_coarse = 8

x_coarse = np.linspace(1, 100, n_coarse)
y_coarse = np.linspace(1, 100, n_coarse)


X_coarse, Y_coarse = np.meshgrid(x_coarse, y_coarse, indexing='ij')

grid_points = np.column_stack((X_coarse.ravel(), Y_coarse.ravel()))

T = 10
Z = np.cos(X_coarse / T) * np.sin(Y_coarse / T)


plt.figure(figsize = [12, 8])
plt.subplot(1,2,1)
plt.title('Original (coarse) data')
plt.imshow(Z, extent=[x_coarse[0], x_coarse[-1], y_coarse[0], y_coarse[-1]], origin='lower')

scale_factor = 10

x_fine = np.logspace(0, 2, n_coarse * scale_factor)
y_fine = np.logspace(0, 2, n_coarse * scale_factor)

X_fine, Y_fine = np.meshgrid(x_fine, y_fine, indexing='ij')

# This USED to be called interp2D but they changed the name in the newest version of scipy:
interpolator_setup = inter.CloughTocher2DInterpolator(grid_points, Z.ravel())
Z_fine = interpolator_setup(X_fine, Y_fine)

plt.subplot(1,2,2)
plt.title(f'Interpolated (upscaled by factor of {scale_factor}) grid')
plt.pcolormesh(x_fine, y_fine, Z_fine)
#plt.imshow(Z_fine, origin='lower')
plt.gca().set_aspect('equal', adjustable='box') 
```

```python
# another cool example:

# Define target image resolution
grid_y, grid_x = np.mgrid[0:100, 0:100]

# Generate 50 random pixel coordinate samples across a 100x100 space
num = 50

known_y = np.random.randint(0, 100, size=num)
known_x = np.random.randint(0, 100, size=num)

# Generate synthetic pixel values (e.g., a smooth wave pattern)
known_values = np.sin(known_x / 10.0) * np.cos(known_y / 10.0)

# Interpolate the unstructured data onto the 100x100 grid
# Options for method: 'nearest', 'linear', 'cubic'
reconstructed_image = inter.griddata(
    (known_y, known_x), known_values, (grid_y, grid_x), method='cubic'
)


plt.figure(figsize = [20, 8])

plt.subplot(1,2,1)
plt.title('original image')
plt.imshow(np.sin(grid_x / 10.0) * np.cos(grid_y / 10.0), cmap='plasma', origin='lower')

plt.subplot(1,2,2)
plt.imshow(reconstructed_image, cmap='plasma', origin='lower')
plt.title("Reconstructed from Scattered Points")

plt.scatter(known_x, known_y, c = 'xkcd:neon green', marker = 'x', s = 10)
plt.colorbar()
plt.show()
```

### scipy minimize:
**scipy.optimize.minimize** is a powerful set of tools for finding the minimum of functions. It's useful when you need to:
- Fit models to data by minimizing a loss function
- Find optimal parameters for a system
- Solve inverse problems
- Optimize under constraints

Let's explore different optimization methods and apply them to a realistic curve-fitting problem with noisy data.

```python
from scipy.optimize import minimize
import matplotlib.pyplot as plt

########################################
# 1) Basic example: minimizing a simple function
########################################

# Define a simple 1D function to minimize
# This is a parabola with minimum at x=3
def simple_function(x):
    return (x - 3)**2 + 2

# Initial guess for where the minimum might be
x_initial = 0.0

# Use the default minimizer (Nelder-Mead for unconstrained scalar functions)
result_simple = minimize(simple_function, x_initial)

print("Simple minimization result:")
print(f"  Minimum found at x = {result_simple.x[0]:.4f}")
print(f"  Function value at minimum = {result_simple.fun:.6f}")
print(f"  Success: {result_simple.success}")
print(20 * '-')

# Visualize the function and the result:
x_vals = np.linspace(-2, 8, 100)
y_vals = simple_function(x_vals)

plt.figure(figsize=(8, 5))
plt.plot(x_vals, y_vals, 'b-', linewidth=2, label='f(x) = (x-3)² + 2')
plt.plot(result_simple.x[0], result_simple.fun, 'r*', markersize=20, label=f'Minimum at x={result_simple.x[0]:.2f}')
plt.scatter([x_initial], [simple_function(x_initial)], color='orange', s=100, marker='o', label=f'Initial guess: x={x_initial}')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.title('Simple 1D Minimization Example')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
########################################
# 2) Comparing different optimization methods
########################################

# Define a 2D function with multiple local minima (Rosenbrock function - classic test function)
def rosenbrock(x):
    return 100 * (x[1] - x[0]**2)**2 + (1 - x[0])**2

# Different optimization methods available in scipy
methods = ['Nelder-Mead', 'BFGS', 'Powell', 'L-BFGS-B']
initial_guess = np.array([-1.0, -1.0])

results_comparison = {}
for method in methods:
    result = minimize(rosenbrock, initial_guess, method=method)
    results_comparison[method] = result
    print(f"Method: {method}")
    print(f"  Minimum: {result.x}")
    print(f"  Function value: {result.fun:.6e}")
    print(f"  Number of iterations: {result.nit}")
    print(f"  Success: {result.success}")
    print()

# Visualize the Rosenbrock function and all optimization paths:
x_range = np.linspace(-2, 2, 300)
y_range = np.linspace(-1, 3, 300)
X, Y = np.meshgrid(x_range, y_range)
Z = rosenbrock(np.array([X, Y]))

plt.figure(figsize=(10, 8))
plt.contourf(X, Y, np.log10(Z + 1), levels=30, cmap='viridis')
plt.colorbar(label='log10(f(x,y) + 1)')

# Plot the known global minimum
plt.plot(1, 1, 'r*', markersize=20, label='Global minimum (1, 1)')

# Plot starting point
plt.plot(initial_guess[0], initial_guess[1], 'o', color='orange', markersize=10, label='Starting point')

# Plot results from each method
colors = ['red', 'cyan', 'lime', 'magenta']
for (method, color) in zip(methods, colors):
    result = results_comparison[method]
    plt.plot(result.x[0], result.x[1], 's', color=color, markersize=8, label=f'{method}: {result.fun:.2e}')

plt.xlabel('x')
plt.ylabel('y')
plt.title('Rosenbrock Function: Different Optimization Methods')
plt.legend(loc='upper left', fontsize=8)
plt.xlim(-2, 2)
plt.ylim(-1, 3)
plt.show()


print(150 * '-')
print('In this example, all methods did basically equally well - in our example, we might have to test different methods to figure out what is the best/fastest.')
print(150 * '-')
```

```python
# This next example is a really useful method that you'll end up using a lot of in science: fitting a function to noisy data.

########################################
# 3) Applied example: Fitting a model to noisy data
########################################

# Generate synthetic data: true model is a Gaussian, plus noise
np.random.seed(42)

# True parameters:
true_amplitude = 5.0
true_mean = 50.0
true_std = 10.0

# Generate x values
x_data = np.linspace(0, 100, 100)

# Generate noisy y values (Gaussian with noise)
y_true = true_amplitude * np.exp(-0.5 * ((x_data - true_mean) / true_std)**2)
noise = np.random.normal(0, 0.3, len(x_data))  # random noise with std=0.3
y_data = y_true + noise

# Define the model to fit
def gaussian_model(x, params):
    amplitude, mean, std = params
    return amplitude * np.exp(-0.5 * ((x - mean) / std)**2)

# we need some way to quantify the differences between our model and our data 
# (basically, some way of putting numbers to the difference between model/data)
# we'll use a sum of squares, a common metric where you square the difference between 
# model and data and then add all them up for all points.
# Define the loss function (sum of squared residuals) that we want to minimize
def loss_function(params):
    y_pred = gaussian_model(x_data, params)
    residuals = y_data - y_pred
    return np.sum(residuals**2)

# Initial guess for parameters (deliberately bad guess)
initial_params = np.array([1.0, 30.0, 5.0])

# Perform optimization
result_fit = minimize(loss_function, initial_params, method='BFGS')

# Extract optimized parameters
fitted_amplitude, fitted_mean, fitted_std = result_fit.x

print("Curve Fitting Result:")
print(f"True parameters:     amplitude={true_amplitude:.2f}, mean={true_mean:.2f}, std={true_std:.2f}")
print(f"Initial guess:       amplitude={initial_params[0]:.2f}, mean={initial_params[1]:.2f}, std={initial_params[2]:.2f}")
print(f"Fitted parameters:   amplitude={fitted_amplitude:.2f}, mean={fitted_mean:.2f}, std={fitted_std:.2f}")
print(f"Loss function value: {result_fit.fun:.6f}")
print(f"Success: {result_fit.success}")
print(20 * '-')

# Generate smooth curves for visualization
x_smooth = np.linspace(0, 100, 200)
y_true_smooth = true_amplitude * np.exp(-0.5 * ((x_smooth - true_mean) / true_std)**2)
y_fitted_smooth = gaussian_model(x_smooth, result_fit.x)

# Plot the results
plt.figure(figsize=(12, 6))

# Left: data with fit
plt.subplot(1, 2, 1)
plt.scatter(x_data, y_data, alpha=0.6, label='Noisy data', s=30)
plt.plot(x_smooth, y_true_smooth, 'g--', linewidth=2, label='True model')
plt.plot(x_smooth, y_fitted_smooth, 'r-', linewidth=2.5, label='Fitted model')
plt.xlabel('x')
plt.ylabel('y')
plt.title('Gaussian Model Fit to Noisy Data')
plt.legend()
plt.grid(True, alpha=0.3)

# Right: residuals
plt.subplot(1, 2, 2)
y_fitted_data = gaussian_model(x_data, result_fit.x)
residuals = y_data - y_fitted_data
plt.scatter(x_data, residuals, alpha=0.6, s=30, color='purple')
plt.axhline(0, color='k', linestyle='--', linewidth=1)
plt.xlabel('x')
plt.ylabel('Residuals (data - fit)')
plt.title('Residuals: Difference Between Data and Fit')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
########################################
# 4) Optimization with bounds and constraints
########################################

# Sometimes you want to constrain parameters to physically reasonable ranges 
# (for us, this will be a guess NEARBY the original (but wrong) header coordinates

# Another example in the previous (Gaussian) example: amplitude must be positive, std must be positive

# Using L-BFGS-B method which allows bounds:
bounds = [
    (0.1, 20.0),      # amplitude: between 0.1 and 20
    (0.0, 100.0),     # mean: between 0 and 100
    (0.5, 50.0)       # std: between 0.5 and 50
]

result_bounded = minimize(loss_function, initial_params, method='L-BFGS-B', bounds=bounds)

print("Bounded Optimization Result:")
print(f"Fitted parameters (bounded):   amplitude={result_bounded.x[0]:.2f}, mean={result_bounded.x[1]:.2f}, std={result_bounded.x[2]:.2f}")
print(f"Loss function value: {result_bounded.fun:.6f}")
print(f"Success: {result_bounded.success}")
print(20 * '-')

# Compare bounded vs unbounded results
print("\nComparison of methods:")
print(f"BFGS (unbounded):   loss = {result_fit.fun:.6f}")
print(f"L-BFGS-B (bounded): loss = {result_bounded.fun:.6f}")

# Visualize parameter space exploration
# Create a grid of parameter values to show the loss landscape
amplitude_range = np.linspace(0.5, 8, 50)
mean_range = np.linspace(20, 80, 50)
loss_grid = np.zeros((len(amplitude_range), len(mean_range)))

# Calculate loss for each combination (keeping std fixed at fitted value)
std_fixed = fitted_std
for i, amp in enumerate(amplitude_range):
    for j, m in enumerate(mean_range):
        loss_grid[i, j] = loss_function([amp, m, std_fixed])

plt.figure(figsize=(10, 7))
plt.contourf(mean_range, amplitude_range, loss_grid, levels=30, cmap='viridis')
plt.colorbar(label='Loss function value')

# Plot the results from both methods
plt.plot(result_fit.x[1], result_fit.x[0], 'r*', markersize=20, label=f'BFGS result')
plt.plot(result_bounded.x[1], result_bounded.x[0], 'c+', markersize=15, markeredgewidth=2, label=f'L-BFGS-B (bounded)')
plt.plot(true_mean, true_amplitude, 'go', markersize=10, label='True parameters')

plt.xlabel('Mean')
plt.ylabel('Amplitude')
plt.title('Loss Function Landscape\n(with std fixed at fitted value)')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```


### Summary: Choosing the Right Optimization Method

- **Nelder-Mead**: General-purpose, no gradient needed, good for noisy/non-smooth functions. Slower but robust.
- **BFGS**: Fast, requires smooth functions, estimates gradient numerically. Good default for well-behaved problems.
- **L-BFGS-B**: Like BFGS but handles bounds on parameters. Use when you have constraints on parameter ranges.
- **Powell**: Similar to Nelder-Mead, doesn't require gradients. Good alternative if Nelder-Mead doesn't converge.

**Key Points:**
- Always define a loss/objective function to minimize (some way to quantify the difference between model and data)
- Provide a reasonable initial guess
- Use bounds (L-BFGS-B) when parameters should/could be constrained
- Check the `success` flag to verify convergence
- Use residual plots to diagnose fit quality
