## 1. Numpy

numpy is a pacakge that allows for efficient creation and manipulation of data-arrays (basically, multi-dimensional data). It's pretty key for any science done with python and has some really useful functionality.

There's a ton of useful cheat-sheets for numpy online, including this one: It looks like a lot of information but don't memorize it, just reference it when you need something. 


 ![alt text](https://ashutoshtripathi.com/wp-content/uploads/2019/02/numpy-cheat-sheet.png)

```python
import numpy as np


# creating an empty data array of shape (5, 5), filled with zero's:
test_array = np.zeros((5, 5))

print(test_array)
print(20 * '-')

# print the shape of the array:
print(test_array.shape)
print(20 * '-')

# The array can be sliced just like lists: 
print(test_array[:3, :3])
print(20 * '-')

# unlike lists, we can do math to the arrays, for instance, multiplying or squaring every entry:

# create an array filled with one's of the same shape as a another array:
test_array2 = np.ones_like(test_array)
print(test_array2)
print(20 * '-')

# do math to it:
print(test_array2 * 5)
print(20 * '-')

# create a list of ascending numbers: format is (start, stop, # steps)
test_array3 = np.linspace(0, 25, 26)
print(test_array3)
print(20 * '-')
```

## 2. matplotlib:

matplotlib is a package used for plotting and vizualizing scientific data! It's pretty flexible - if you can imagine a plot, you can make it! But it has some more basic functions like plot, scatter (for a scatterplot), imshow, and hist (for a histogram).

If you want to do anything specific or change details of the plot, you can look up the documentation (for instance, for [plt.plot](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html)) or google it.

It also works AMAZINGLY together with numpy arrays! 

```python
# basic matplotlib examples
import matplotlib.pyplot as plt

########################################
# 1) simple line plot with plt.plot:

# create some test data:
x = np.linspace(0, 10, 100)
y = np.sin(x)

# create an empty plot 
plt.figure(figsize=(6, 3))
# plot the data! x vs. y:
plt.plot(x, y, color='blue', marker='o', markersize=4, label='sin(x)')

# add text to the axes/title:
plt.title('Line plot with plt.plot')
plt.xlabel('x')
plt.ylabel('sin(x)')

# add a legend:
plt.legend()

# add gridlines:
plt.grid(True)

# show it! (this is run automatically in jupyter notebooks, but you'll need this command when running NOT in jupyter notebooks)
plt.show()

########################################

# 2) scatter plot with plt.scatter:
# create random x/y data for individual points:
x_scatter = np.random.rand(50)
y_scatter = np.random.rand(50)
# create a color value and size for each point:
colors = np.random.rand(50)
sizes = 100 * np.random.rand(50)

plt.figure(figsize=(6, 4))
# scatter plots show each point separately instead of connecting them:
plt.scatter(x_scatter, y_scatter, c=colors, s=sizes, alpha=0.7, cmap='viridis')

# label the chart and axes so the plot is easy to read:
plt.title('Scatter plot with plt.scatter')
plt.xlabel('x values')
plt.ylabel('y values')

# add a colorbar to explain the mapping from color to value:
plt.colorbar(label='color scale')
plt.show()

########################################

# 3) image display with plt.imshow:
# create a 2D array of values to display as an image:
image_data = np.random.rand(10, 10)

plt.figure(figsize=(4, 4))
# imshow renders the 2D array as a colored image:
plt.imshow(image_data, cmap='viridis', interpolation='nearest')

# add a title and a colorbar so values are easier to interpret:
plt.title('Image display with plt.imshow')
plt.colorbar(label='pixel intensity')

# hide the axes to focus on the image itself:
plt.axis('off')
plt.show()

########################################

# 4) histogram with plt.hist:
# generate random data to summarize in bins:
values = np.random.randn(1000)

plt.figure(figsize=(6, 4))
# hist groups the values into bins and draws their counts:
plt.hist(values, bins=30, color='skyblue', edgecolor='black')

# add labels and title so the distribution is clear:
plt.title('Histogram with plt.hist')
plt.xlabel('value')
plt.ylabel('frequency')
plt.show()
```

# astropy:

astropy is package used for astronomy data - it's a collection of functions that are useful and commonly used in astronomy data, and contains a TON of functionality - we'll only be using some small parts of it:

1. using astropy units for converting/handling numbers with units
2. using astropy.io.fits to read in/handle fits data and headers

```python
# astropy examples: units and FITS handling
from astropy import units as u
from astropy.io import fits

########################################
# 1) using astropy units:
# attach units to plain numbers so arithmetic keeps track of units automatically
length = 10 * u.meter  # 10 meters
print('length:', length)

# convert between compatible units easily:
print('length in cm:', length.to(u.cm))

# combine units in expressions (speed * time -> distance):
speed = 30 * u.km / u.s  # 30 kilometers per second
time = 10 * u.s           # 10 seconds
# resulting quantity carries the correct unit (km):
distance = speed * time
print('distance (km):', distance.to(u.km))

# unit-aware arrays (wrap a numpy array with a unit):
speeds = np.array([10., 20., 30.]) * u.m / u.s
print('speeds:', speeds)
# you can compute stats and convert results to other units:
print('mean speed in km/h:', speeds.mean().to(u.km / u.h))


print(50 * '-')
########################################
# 2) reading and writing FITS files with astropy.io.fits:

test_fits = fits.open('/Users/jamescrowley/Documents/summer_2026/research/pid_4_62/IHFDSO/VISP_2026_05_13T23_56_31_823023_00589100_I_IHFDSO_L1.fits')[1]

test_fits_header = test_fits.header
test_fits_data = test_fits.data
# add a small header keyword to demonstrate accessing header info:
# The shape of the data:
print(test_fits_data.shape)

# Accesssing one of the FITS header keywords:
print(test_fits_header['CUNIT3'])

plt.figure(figsize = [20, 8])

# Plotting a spectra from a single pixel: 
plt.subplot(1,2,1)
plt.plot(test_fits_data[0, :, 10])
plt.axvline(10, c = 'r', ls = '--')

# Plotting the average of the first 20 wavelength indexes across ALL pixels across the slit
plt.subplot(1,2,2)
plt.plot(test_fits_data[0, 10, :])
plt.plot(np.nanmean(test_fits_data[0, :20, :], axis = 0))
plt.ylim(0.8, 1.2)
plt.show()
```
